```python
# CNN Handwritten Digit Classification using MNIST

import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt



(x_train, y_train), (x_test, y_test) = datasets.mnist.load_data()

print("Training data shape:", x_train.shape)
print("Testing data shape:", x_test.shape)


x_train = x_train / 255.0
x_test = x_test / 255.0

x_train = x_train.reshape(-1, 28, 28, 1)
x_test = x_test.reshape(-1, 28, 28, 1)


model = models.Sequential([

    # Convolutional layer
    layers.Conv2D(
        32,
        (3, 3),
        activation='relu',
        input_shape=(28, 28, 1)
    ),

    # Reduce spatial size
    layers.MaxPooling2D((2, 2)),

    # Second convolutional layer
    layers.Conv2D(
        64,
        (3, 3),
        activation='relu'
    ),

    # Reduce spatial size again
    layers.MaxPooling2D((2, 2)),

    # Convert feature maps into a 1D vector
    layers.Flatten(),

    # Fully connected layer
    layers.Dense(64, activation='relu'),

    # Output layer - 10 digits
    layers.Dense(10, activation='softmax')
])



model.summary()

model.compile(
    optimizer='adam',
    loss='sparse_categorical_crossentropy',
    metrics=['accuracy']
)

history = model.fit(
    x_train,
    y_train,
    epochs=5,
    validation_split=0.1
)


test_loss, test_accuracy = model.evaluate(
    x_test,
    y_test
)

print("Test Accuracy:", test_accuracy)

predictions = model.predict(x_test)

# Get the digit with highest probability
predicted_digit = predictions.argmax(axis=1)

print("Actual digit:   ", y_test[0])
print("Predicted digit:", predicted_digit[0])

plt.imshow(x_test[0].reshape(28, 28), cmap='gray')
plt.title(
    f"Actual: {y_test[0]} | Predicted: {predicted_digit[0]}"
)
plt.axis('off')
plt.show()
```
